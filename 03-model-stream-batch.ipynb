```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv

# 加载配置文件
load_dotenv(override=True)

GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
GOOGLE_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

# 確保環境變數有 GOOGLE_API_KEY，或是直接在這裡把 key 傳進 api_key 參數
model = init_chat_model(
    model="gemma-4-31b-it",
    model_provider="openai",  # 關鍵：強制使用 OpenAI 的連線邏輯
    api_key=os.getenv("GOOGLE_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/" # 指向 Google
)
print(model)
```

```python


for clunk in model.stream("幫我解釋一下什麼是人工智能"):
    print(clunk.text,end="",flush=True)
```

# 批量調用

## 舉例1:


```python
from rich import print as rprint
messages = [
    "您好，你是誰?",
    "2 + 3 * 5 = ?",
    "美國首都在哪裡?"
]

responses = model.batch(messages)

for res  in responses:
    rprint(res)


```

## 舉例2:按照完成的順序接收

```python
messages = [
    "您好，你是誰?",
    "2 + 3 * 5 = ?",
    "美國首都在哪裡?"
]

responses = model.batch_as_completed(messages)

for res  in responses:
    rprint(res)
```
