# ChatPromptTemplate的使用

## ChatPromptTemplate的實例化

### 1.1 from_messages()

EX:

```python
from langchain_core.prompts import ChatPromptTemplate


ChatPromptTemplate = ChatPromptTemplate.from_messages(
    [
        ("system","你是一個有好的AI助手，你的名子叫{name}"),
        ("human","你好，最近怎麼樣?"),
        ("ai","我很好，謝謝"),
        ("human","{user_input}")
    ]
)

# 調用
res = ChatPromptTemplate.invoke({
    "name":"小智","user_input":"2+2=?"
})

print(res)


```
